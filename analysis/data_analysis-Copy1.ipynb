```python
# 欠損値を確認
print(df.isnull().sum())

# annotation_num と各レベルの投票数が一致するか確認
level_columns = [
    "Not Toxic",
    "Hard to Say",
    "Toxic",
    "Very Toxic"
]

print(
    "投票数の合計とannotation_numが一致しない行数:",
    (df[level_columns].sum(axis=1) != df["annotation_num"]).sum()
)
```

```python
# 各レベルの重み
level_weights = {
    "Not Toxic": 0,
    "Hard to Say": 1,
    "Toxic": 2,
    "Very Toxic": 3
}

# 重み付き平均を計算
df["level_score"] = sum(
    df[column] * weight
    for column, weight in level_weights.items()
) / df["annotation_num"]

# 四捨五入して整数に変換
# np.round()ではなく、0.5を切り上げる方法を使用
df["level"] = np.floor(df["level_score"] + 0.5).astype(int)

# 確認
df[["text", "Not Toxic", "Hard to Say",
    "Toxic", "Very Toxic", "annotation_num",
    "level_score", "level"]].head(20)
```

```python
# カテゴリ列
category_columns = [
    "category_卑語",
    "category_差別",
    "category_迷惑行為",
    "category_猥褻",
    "category_出会い・プライバシー侵害",
    "category_違法行為",
    "category_偏向表現"
]

def determine_category(row):
    # 各カテゴリの投票数
    votes = row[category_columns]

    # 最大投票数
    max_votes = votes.max()

    # すべて0なら該当カテゴリなし
    if pd.isna(max_votes) or max_votes == 0:
        return None

    # 最大投票数のカテゴリをすべて取得
    winners = votes[votes == max_votes].index.tolist()

    # category_ を取り除いて格納
    return "|".join(
        category.replace("category_", "")
        for category in winners
    )

df["category"] = df.apply(determine_category, axis=1)

# 確認
df[["text", "category"]].head(20)
```

```python
# いったん欠損値で初期化
df["classification"] = pd.NA

# 自動で決められるレベルを設定
df.loc[df["level"] == 0, "classification"] = 0
df.loc[df["level"].isin([2, 3]), "classification"] = 1

# 整数型を保ち、未設定値を許容
df["classification"] = df["classification"].astype("Int64")

# 確認
print(df.groupby(["level", "classification"], dropna=False).size())
```

```python
# 手動確認が必要な行
hard_to_say_df = df.loc[
    df["level"] == 1,
    ["id", "text", "level_score", "level", "classification"]
].copy()

display(hard_to_say_df)
```

```python
# 未設定の分類が残っていないか確認
print("classificationの未設定件数:",
      df["classification"].isna().sum())

# 分類ごとの件数
print(df["classification"].value_counts(dropna=False))
```

```python
# 出力先
output_path = "../data/processed/toxic_dataset_1.csv"

# 必要な列を選択
output_columns = [
    "id",
    "text",
    "level_score",
    "level",
    "classification",
    "category",
    "annotation_num"
]

df[output_columns].to_csv(
    output_path,
    index=False,
    encoding="utf-8-sig"
)

print(f"保存しました: {output_path}")
```

## 学習

```python
from sklearn.model_selection import train_test_split

df = pd.read_csv("../data/processed/toxic_dataset_only_class.csv")

train_df, temp_df = train_test_split(
    df,
    test_size=0.2,
    random_state=42,
    stratify=df["classification"]
)

valid_df, test_df = train_test_split(
    temp_df,
    test_size=0.5,
    random_state=42,
    stratify=temp_df["classification"]
)

print("train:", len(train_df))
print("valid:", len(valid_df))
print("test :", len(test_df))
```

```python
from datasets import Dataset

train_dataset = Dataset.from_pandas(
    train_df[["text", "classification"]]
)

valid_dataset = Dataset.from_pandas(
    valid_df[["text", "classification"]]
)

test_dataset = Dataset.from_pandas(
    test_df[["text", "classification"]]
)

print("="*80)
print(train_dataset)
print(train_dataset[0])
print("="*80)
print(valid_dataset)
print(valid_dataset[0])
print("="*80)
print(test_dataset)
print(test_dataset[0])
```

```python
from transformers import AutoTokenizer

MODEL_NAME = "cl-tohoku/bert-base-japanese-v3"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
```

```python
def tokenize_function(examples):
    return tokenizer(
        examples["text"],
        padding="max_length",
        truncation=True,
        max_length=128
    )
```

```python
tokenized_train = train_dataset.map(
    tokenize_function,
    batched=True
)

tokenized_valid = valid_dataset.map(
    tokenize_function,
    batched=True
)

tokenized_test = test_dataset.map(
    tokenize_function,
    batched=True
)

print(tokenized_train[0])
print(tokenized_valid[0])
print(tokenized_test[0])
```

```python
text = train_dataset[0]["text"]

print("元の文章:")
print(text)

print("\nトークン:")
print(tokenizer.tokenize(text))

print("\nID:")
print(tokenizer.convert_tokens_to_ids(tokenizer.tokenize(text)))

print("\nTokenizerの出力:")
print(tokenizer(
    text,
    padding="max_length",
    truncation=True,
    max_length=128
))
```

## BERTでの学習

```python
from transformers import AutoModelForSequenceClassification
```

```python
model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=2
)
```

```python
def rename_label(dataset):
    return dataset.rename_column(
        "classification",
        "labels"
    )

tokenized_train = rename_label(tokenized_train)
tokenized_valid = rename_label(tokenized_valid)
tokenized_test = rename_label(tokenized_test)
```

```python
print(tokenized_train[0])
```

```python
from transformers import TrainingArguments
import numpy as np
from sklearn.metrics import accuracy_score, precision_recall_fscore_support
from transformers import Trainer
```

### 重み付け無しの場合

```python
training_args = TrainingArguments(
    output_dir="./models/toxic-bert",
    eval_strategy="epoch",
    save_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=3,
    weight_decay=0.01,
    # logging_dir="./logs",
    logging_steps=50,
    load_best_model_at_end=True
)

training_args_1 = TrainingArguments(
    output_dir="./models/toxic-bert-tuned",
    eval_strategy="epoch",
    save_strategy="epoch",

    learning_rate=1e-5,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,

    num_train_epochs=5,
    weight_decay=0.01,

    logging_steps=10,
    load_best_model_at_end=True,

    metric_for_best_model="eval_f1",
    greater_is_better=True
)

def compute_metrics(eval_pred):
    logits, labels = eval_pred

    predictions = np.argmax(logits, axis=-1)

    precision, recall, f1, _ = precision_recall_fscore_support(
        labels,
        predictions,
        average="binary"
    )

    accuracy = accuracy_score(
        labels,
        predictions
    )

    return {
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1
    }
```

```python
trainer = Trainer(
    model=model,
    args=training_args_1,
    train_dataset=tokenized_train,
    eval_dataset=tokenized_valid,
    compute_metrics=compute_metrics
)

trainer.train()
```

```python
test_result = trainer.evaluate(
    tokenized_test
)

print(test_result)
```

### クラス重みづけを追加

```python
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

classes = np.array([0, 1])

class_weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=train_df["classification"].astype(int)
)

print(class_weights)
```

```python
import torch
from transformers import Trainer

class WeightedTrainer(Trainer):
    def compute_loss(
        self,
        model,
        inputs,
        return_outputs=False,
        num_items_in_batch=None
    ):
        labels = inputs.pop("labels")
        outputs = model(**inputs)
        logits = outputs.logits

        weights = torch.tensor(
            class_weights,
            dtype=torch.float32,
            device=logits.device
        )

        loss_function = torch.nn.CrossEntropyLoss(
            weight=weights
        )

        loss = loss_function(logits, labels)

        return (loss, outputs) if return_outputs else loss
```

```python
training_args = TrainingArguments(
    output_dir="./models/toxic-bert-weighted",
    eval_strategy="epoch",
    save_strategy="epoch",
    learning_rate=1e-5,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=5,
    weight_decay=0.01,
    logging_steps=10,
    load_best_model_at_end=True,
    metric_for_best_model="eval_f1",
    greater_is_better=True
)

trainer = WeightedTrainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train,
    eval_dataset=tokenized_valid,
    compute_metrics=compute_metrics
)

trainer.train()
```

```python
test_result = trainer.evaluate(
    tokenized_test
)

print(test_result)
```

```python

import numpy as np
import torch
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# 検証データで予測する（テストデータは使わない）
predictions = trainer.predict(tokenized_valid)
logits = predictions.predictions
true_labels = predictions.label_ids

# クラス1（不適切）の予測確率を計算
probabilities = torch.softmax(
    torch.tensor(logits),
    dim=1
).numpy()[:, 1]

# 複数の閾値を比較
for threshold in [0.3, 0.4, 0.5, 0.6, 0.7]:
    pred_labels = (
        probabilities >= threshold
    ).astype(int)

    print(f"\nThreshold: {threshold}")
    print("Confusion matrix:")
    print(confusion_matrix(
        true_labels, pred_labels, labels=[0, 1]
    ))
    print(f"Accuracy : {accuracy_score(true_labels, pred_labels):.3f}")
    print(f"Precision: {precision_score(true_labels, pred_labels, zero_division=0):.3f}")
    print(f"Recall   : {recall_score(true_labels, pred_labels, zero_division=0):.3f}")
    print(f"F1       : {f1_score(true_labels, pred_labels, zero_division=0):.3f}")
```

```python
# 検証データの予測確率と正解ラベルを確認
for i, (prob, label) in enumerate(
    zip(probabilities, true_labels)
):
    print(
        f"Index={i}, "
        f"正解={label}, "
        f"不適切確率={prob:.4f}"
    )
```
